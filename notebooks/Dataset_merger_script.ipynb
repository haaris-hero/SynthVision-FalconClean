```python
import requests
import zipfile
import os
import shutil
import io

# Define final directories
images_dir = '/content/images'
labels_dir = '/content/labels'
os.makedirs(images_dir, exist_ok=True)
os.makedirs(labels_dir, exist_ok=True)

def copy_and_rename_unique(source_folder, destination_folder, prefix):
    os.makedirs(destination_folder, exist_ok=True)
    existing_files = set(os.listdir(destination_folder))

    for filename in os.listdir(source_folder):
        source_path = os.path.join(source_folder, filename)
        if os.path.isfile(source_path):
            unique_name = prefix + filename
            base_name, extension = os.path.splitext(filename)
            counter = 1

            while unique_name in existing_files:
                unique_name = f"{prefix}{base_name}_{counter}{extension}"
                counter += 1

            destination_path = os.path.join(destination_folder, unique_name)
            shutil.copy2(source_path, destination_path)
            existing_files.add(unique_name)
            print(f"Copied: {filename} → {unique_name}")

def process_zip(url, set_id):
    prefix = f"set{set_id}_"

    # Download and extract in memory
    response = requests.get(url)
    with zipfile.ZipFile(io.BytesIO(response.content)) as zip_ref:
        zip_ref.extractall(f"unzipped_output_{set_id}")
    print(f"Download and extraction complete for set {set_id}.")

    # Process files
    output_dir = os.path.join(f"unzipped_output_{set_id}", "Output")
    if not os.path.exists(output_dir):
        print(f"Output directory not found for set {set_id}.")
        shutil.rmtree(f"unzipped_output_{set_id}")  # Clean up
        return

    timestamp_dirs = [d for d in os.listdir(output_dir) if os.path.isdir(os.path.join(output_dir, d))]
    if len(timestamp_dirs) != 1:
        print(f"Unexpected number of timestamp directories for set {set_id}.")
        shutil.rmtree(f"unzipped_output_{set_id}")  # Clean up
        return

    base_path = os.path.join(output_dir, timestamp_dirs[0])

    # Process images and labels
    for data_type in ["train", "val"]:
        for data_class in ["images", "labels"]:
            source_path = os.path.join(base_path, data_type, data_class)
            dest_dir = images_dir if data_class == "images" else labels_dir
            if os.path.exists(source_path):
                copy_and_rename_unique(source_path, dest_dir, prefix + f"{data_type}_")

    # Clean up unzipped directory
    shutil.rmtree(f"unzipped_output_{set_id}")
    print(f"Cleaned up extraction directory for set {set_id}")

# List of URLs to process
urls = [
'https://storage.googleapis.com/theresult42/098/09837506-9566-473d-bf76-9d154492f825/output.zip?GoogleAccessId=cloud-api%40gentle-impulse-371419.iam.gserviceaccount.com&Expires=1757112079&Signature=xYh14fGCJM7pjATxzkVjW%2BcHznR%2BQEV1POyUbbJZCAkqy2TZfYKJXi7EPvR78fW%2BWKeTO6hMiw5w1PCrHq2tjVVScjcciD0%2FExACvzybRWQK4mm9gWPrmxZe81E08ttg2kR1gXhj2ZxTAOQRTX9B0%2FLr%2BwNvFSiYuIEsZFOfNDuMlfmBeyCgiBsMIc%2FI9Bvbn0PViScz9BY5kaFdzg0rrwSop5Rjw%2FRA0pipic5KeBhFrED7VfA8C17qSbseKw5NBITzIX2bByD8cfR7RsCKVDC%2Br0DVdfuZDSoC8YU1ZjtTabMk0%2Flnz%2BLfHuiS1rPAe%2BI10ef%2FyfOkeE%2B2wxG%2BqQ%3D%3D&response-content-disposition=attachment%3B%20filename%3D%22output.zip%22',
'https://storage.googleapis.com/theresult42/aeb/aebc8fbf-b897-4f7f-a4fc-95e936e37940/output.zip?GoogleAccessId=cloud-api%40gentle-impulse-371419.iam.gserviceaccount.com&Expires=1757112103&Signature=pCnWo2fXhBGvGIyvw65DR1kqu7aX14VoFdn1lp7%2BzaZAGqLBS1FaTWypa7eStaKosGjpatpjkK%2FO%2BaDEhOOgXFd8uuZyFLjEk2iMo9L%2F%2FLrzag2kPFlC8t1x5aCZ%2BXzQx9HWqh5wJ3s2s%2Fdqqw4cdto3rPioHuHUYVBKr%2BMVbi9HNNhGHAZMw5bLwIRwffKcHo4PQhVkweat%2B1jcpwWtOpmbv%2BAlAJUW58CPOUmvUW1pQDpva95thGwNZDaPYQAz%2Boye2Sgv6fJt3BjUV9mgebOj12CbI6I8zHe0OnIl9g38aXgyuNKBHCMJuGH3h7X4u6XFQu2oBgne25By8sXlYg%3D%3D&response-content-disposition=attachment%3B%20filename%3D%22output.zip%22',
]

# Process each URL
for i, url in enumerate(urls, 1):
    process_zip(url, i)
```
